# 0. 실행 환경 확인

Colab 서버에서 실행된다. 계산은 하지 않고 위치·파일·GPU·버전만 확인한다.

```python
import os, sys, platform, subprocess, importlib.util
from pathlib import Path

print('python      :', sys.version.split()[0], platform.platform())
print('작업 폴더    :', os.getcwd())
print('작업 폴더 내용:', sorted(os.listdir('.'))[:30])
print('/content 존재:', Path('/content').exists())

# 내 컴퓨터의 저장소가 서버에서 보이는지: 위로 올라가며 src/와 config/experiment.yaml을 찾는다
found = next((p for p in [Path.cwd(), *Path.cwd().parents] if (p / 'src').is_dir() and (p / 'config' / 'experiment.yaml').is_file()), None)
print('저장소 루트   :', found)

print(subprocess.run(['nvidia-smi', '--query-gpu=name,memory.total,driver_version', '--format=csv'], capture_output=True, text=True).stdout or 'nvidia-smi 없음')

for m in ['numpy', 'pandas', 'pyarrow', 'scipy', 'sklearn', 'xgboost', 'lightgbm', 'yaml']:
    if importlib.util.find_spec(m):
        mod = __import__(m)
        print(f'{m:9s}', getattr(mod, '__version__', '?'))
    else:
        print(f'{m:9s}', '설치 안 됨')
```
